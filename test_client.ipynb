```python
import socket

port = 5000
hostname = '127.0.0.1'

client = socket.socket()
try:
    client.settimeout(5.0)
    client.connect((hostname, port))
    data = client.recv(100).decode('utf-8')
    print(data)
    client.close()
except socket.timeout:
    client.close()
except Exception as e:
    print(f"错误：{e}")
```
