# 循环与分支结构：

```python
i = 0
while i<20:
    i+=2
print(i)

# 求最大连续子段和
def large_cont_sum(arr):
    cur = 0
    ans = 0
    for i in arr: # 循环与分支结构
        # 缩进表达程序块
        if cur + i >= 0:
            cur = cur + i
            ans = max(ans, cur)
        # elif ...:
        else:
            cur = 0
    return ans

ans = large_cont_sum([1,2,-1,3,4,10,10,-10,-1])
print(ans)
```

在Python中，依托循环结构与生成表达式惰性产生数据的对象被称为生成器：generator  
容器（container）类（如：list，string，set，tuple，dict…）都内建有一个操作函数来返回一个迭代器（iterator）。  
迭代器Iterator对象表示的是一个数据流，Iterator对象可以被next()函数调用并不断返回下一个数据。  
迭代器不但可以作用于for循环，还可以被next()函数不断调用并返回下一个值，直到最后抛出StopIteration错误表示无法继续返回下一个值。  


```python
# 迭代器（容器）类型
# 列表list、元组tuple、集合set、词典dict——容器container
# 遍历计算容器内元素：
a = list("12345")
for i in a:
    print(int(i) * 8)

# 迭代器Iterator对象
x = iter(a)
print(type(x))
print(next(x), next(x))
b = []
for i in x:
    b.append(i)
print(b)
```

```python
# 列表实现快速排序算法
def quicksort(arr):
    if len(arr) <= 1:  # 递归返回条件
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]  # 生成小于pivot的左列表
    middle = [x for x in arr if x == pivot]  # middle列表，支持有重复值情况
    right = [x for x in arr if x > pivot]  # 生成大于pivot的右列表
    return quicksort(left) + middle + quicksort(right)  # 返回一个新的列表对象

print(quicksort([3, 6, 8, 10, 1, 2, 1]))
```

```python
# 用函数实现一个 generator
# —— yield关键字
# 带有 yield 的函数不再是一个普通函数，Python 解释器会将其视为一个 generator，
# 调用 fib(5) 不会执行 fib 函数，而是返回一个 iterable 对象。
# 每次next调用，执行到yield位置函数返回next元素
def fib(max = 10): # 设置缺省参数值
    n, a, b = 0, 0, 1
    while n < max:
        yield b
        a, b = b, a + b
        n = n + 1

f = fib(8) # <generator object fib at 0x000002084C185BA0>
for i in f:
    print(i, end=",")
print()

f = fib(5)
print(next(f))
print(next(f))
print(next(f))

# 生成器函数：派生多个生成器实例
f1 = fib(5)
f2 = fib(9)
print(next(f1),next(f1),next(f1),next(f2))
print([i for i in f1])
print([i for i in f2]) # f1, f2 相互独立
```

```python
# map()
# map()函数接收两个参数，一个是函数，一个是Iterable的对象，
# map将传入的函数依次作用到序列的每个元素，并把结果作为新的Iterator返回。
def f(x):
    return x * x
r = map(f, [1, 2, 3])
print(list(r))
```

```python
# filter()
# 过滤序列，filter()接收一个函数和一个序列。
# 把传入的函数依次作用于每个元素，然后根据返回值是True还是False决定保留还是丢弃该元素。
def is_odd(n):
    return n % 2 == 1
r = list(filter(is_odd, [1, 2, 3]))
print(r)
print(list(filter(lambda x: x % 2 == 1, [1, 2, 3])))
```

```python
# 递归函数+lambda表达式实现快排
def qsort(a):
    if len(a) <= 1:
        return a
    else:
        return (qsort(list(filter(lambda x: x <= a[0], a[1:])))
               + [a[0]] # a[0]是哨兵
               + qsort(list(filter(lambda x: x > a[0], a[1:]))))
print(qsort([3, 6, 8, 10, 1, 2, 1]))
```

```python
# zip()
# 用于将可迭代的对象作为参数， 将对象中对应的元素打包成一个个元组，然后返回由这些元组组成的列表
print(list(zip('abc', 'bcd')))
s = [[i, j] for i, j in zip('abc', 'bcd')]
print(s)

# 列表生成式和zip()实现矩阵乘法
X = [[12, 7, 3],    # 3x3矩阵
     [4, 5, 6], 
     [7, 8, 9]]

Y = [[5, 8, 1, 2],  # 3x4矩阵
     [6, 7, 3, 0], 
     [4, 5, 9, 1]]

# 注意，使用sum()时如果前面有自己定义的sum变量会冲突，这里使用明确的内置函数builtins.sum()
import builtins
result = [[builtins.sum(a*b for a, b in zip(X_row, Y_col))
          for Y_col in zip(*Y)] # 先解包再zip相当于对Y进行转置
          for X_row in X]
for r in result:
    print(r)

print(*Y)
print(list(zip(*Y)))
```

```python
# 在 Python3 中，reduce() 函数已经被从全局名字空间里移除了，它现在被放置在 functools 模块里，
# 如果想要使用它，则需要通过引入 functools 模块来调用 reduce() 函数。
from functools import reduce

def str2num(s):
    return float(s) if '.' in s else int(s)

def calc(exp):
    ss = exp.split('+')
    ns = map(str2num, ss)
    return reduce(lambda acc, x: acc + x, ns)
    # reduce() 对一个序列进行累积计算，把前两个元素的计算结果与下一个元素继续计算。

r = calc('99 + 88 + 7.6')
print('99 + 88 + 7.6 =', r)

def do_sum(inp):
    # return sum(map(int, inp.split()))
    return reduce(lambda x,y:x+y, map(int, inp.split()))

print(do_sum('1 2 3 4 5'))
```

```python
# 变量的作用域 local、nonlocal & global
# Python引用变量的顺序为：当前作用域局部变量->外层作用域变量->当前模块中的全局变量->Python内置变量
# global关键字可以用在任何地方，包括最上层函数中和嵌套函数中，即使之前未定义该变量，global修饰后也可以直接使用
# nonlocal关键字只能用于嵌套函数中，并且外层函数中定义了相应的局部变量
def scope_test():
    def do_local():
        spam = "local spam No.1"  # 1号 局部变量
    def do_nonlocal():
        nonlocal spam # 外层变量
        spam = "nonlocal spam No.2"  # 2号
    def do_global():
        global spam # 全局变量
        spam = "global spam No.3"  # 3号
    spam = "test spam"  # 外层函数的局部变量（2号）
    do_local()
    print("After local assignment:", spam)  # 打印 2号 spam
    do_nonlocal()
    print("After nonlocal assignment:", spam)
    do_global()
    print("After global assignment:", spam)

try:
    print(spam)
except:
    print("not defined")
scope_test()
print("In global scope:", spam)  # 注意缩进 引用全局变量
```

```python
# 函数闭包：内部定义了一个函数，然后把该函数作为返回值
def print_msg(msg):
    hi = 'Hi,'
    
    def printer(x):
        nonlocal hi    # 不声明则认为是局部变量，在使用前没有赋值
                        # UnboundLocalError: local variable 'hi' referenced before assignment
        hi += msg
        print(hi + x)    # 引用了外部函数的参数
    
    return printer    # 生成并返回一个函数实例

# We execute the function
f1 = print_msg('morning!')
f2 = print_msg('nice day!')

f1('sir')
f2('madam')
```

```python
# 函数闭包还可以实现类似函数模板的功能
def inc(x):
    return x + 1

def dec(x):
    return x - 1

def make_operate_of(func):  # 要求输入一个函数
    def operate(x):  # 要求输入一个参数
        return func(x)
    return operate

addone = make_operate_of(inc)    # 输出一个指定的计算模式
minusone = make_operate_of(dec)
print(addone(2))
print(minusone(2))

# 可以轻松创建其他操作
double = make_operate_of(lambda x: x * 2)
square = make_operate_of(lambda x: x ** 2)
print(double(5))   # 10
print(square(3))   # 9
```

```python
# 偏函数（部分函数 partial function）
from functools import partial

# 原始函数
def multiply(x, y, z):
    return x * y * z

# 创建偏函数：固定第一个参数为2
double = partial(multiply, 2)
x2x5 = partial(multiply, 2, 5)

print(double(3, 4))    # 相当于 multiply(2, 3, 4) = 24
print(x2x5(6))    # 相当于 multiply(2, 5, 6) = 60
```

```python
# 函数装饰器：一般用于抽取共性操作作为功能切片，对一类函数进行包装
def smart_divide(func):  # 对除法操作进行安全检查
    def inner(a, b):
        print("I am going to divide", a, "and", b)
        if b == 0:
            print("除数不能为0")
            return
        return func(a, b)
    return inner

sdivide = smart_divide(lambda a, b: a / b) # 函数闭包做法
sdivide(4, 3)

@smart_divide # @ 号加装饰器函数名放在函数定义的前面，表明定义了一个装饰后的函数实例
def divide(a, b):
    return a / b

divide(4, 3)
divide(4, 0)
```

```python
# 使用装饰器计算函数运行的时间并输出
import time
def log_time(func):
    def inner(*args):
        arg_str = ','.join(list(map(str, args)))
        now = time.time()
        ret = func(*args)
        interval = time.time() - now
        print("{}({}):[{}]s".format(func.__name__, arg_str, interval))
        return ret
    return inner

@log_time
def time_cosume(*args):
    return sum(range(1,1000000))
print("result = ",time_cosume(1,2,3,4,[1,2,3,4],{1:[1,2]}))
```

```python
# 通过函数闭包对装饰器进行定制
# 带参数的装饰器
def specify(req):
    def decorator(func):  # 装饰器函数
        def dechouse():
            print(req + "木地板、", "吊顶", end="")  # end替换换行符
            func()
        return dechouse
    return decorator  # 函数闭包，返回定制的装饰器函数

@specify('高档')  # 通过参数实现定制化的装饰器
def house():
    print("房子")

@specify('普通')  # 通过参数实现定制化的装饰器
def officeroom():
    print("办公室")

house() # 等同于specify("高档")(house_)()
officeroom()
```

```python
# 装饰器可以嵌套（装饰器链）
def star(func):
    def inner(*args):
        print("*" * 30)    # 重复30次
        func(*args)
        print("*" * 30)
    return inner

def percent(func):
    def inner(*args):
        print("%" * 30)
        func(*args)
        print("%" * 30)
    return inner

@star
@percent   # 按下层对应的是更内层
def printer(msg):
    print(msg)

printer("今天沙尘太大啦！！！")
```

```python
# 装饰器实现记忆化存储
def memorize_factorial(f):  # 输入被装饰的操作函数f
    memory = {}    # 所有被装饰的函数实例共享一个公共缓存（记忆化存储）

    def inner(num):
        if num not in memory:    # 查询缓存
            memory[num] = f(num)    # 递归调用操作函数f，返回结果进行缓存写入
            print("%d not in memory" % (num), end=" ")
        return memory[num]    # 如果缓存有内容，则直接返回结果，对递归函数进行优化
    return inner

@memorize_factorial
def facto(num):
    if num == 1:
        return 1
    else:
        return num * facto(num-1)

print(facto(4))
print(facto(5))
print(facto(3))
```

```python
def memoise(func):
    memory = {}   
    def inner(nums):
        nums = tuple(nums)
        if nums not in memory:
            memory[nums] = func(nums)
        return memory[nums]
    return inner

@memoise
def lis(nums:list):
    nums = tuple(nums)
    if len(nums) == 0:
        return 0
    else:
        return max(lis(nums[1:]), lis(list(filter(lambda x: x >= nums[0], nums[1:]))) + 1)

# lis函数的作用是求序列的最大上升子序列的长度
print(lis([4,9,8,7,3,1,3,5,6,7,8]))
```

### python引用计数
在一些代码中，如果存在一些变量但是没有用，会造成内存空间，因此叫做垃圾，所以要回收。

引用计数增加的场景有：
* 对象被创建并赋值给某个变量，比如：a = 'ABC'
* 变量间的相互引用（相当于变量指向了同一个对象），比如：b=a
* 变量作为参数传到函数中。比如：ref_method(a)，
* 将对象放到某个容器对象中(列表、元组、字典)。比如：c = [1, a, 'abc']

引用计数减少的场景有：
* 当一个变量离开了作用域，比如：函数执行完成时，执行方法前后的引用计数保持不变，这就是因为方法执行完后，对象的引用计数也会减少，如果在方法内打印，则能看到引用计数增加的效果。
* 对象的引用变量被销毁时，比如del a或者del b。注意如果del a，再去获取a的引用计数会直接报错。
* 对象被从容器对象中移除，比如：c.remove(a)
* 直接将整个容器销毁，比如：del c
* 对象的引用被赋值给其他对象，相当于变量不指向之前的对象，而是指向了一个新的对象，这种情况，引用计数肯定会发生改变。(排除两个对象默认引用计一致的场景)


```python
import sys

# 进入函数会使参数的引用计数+2
# 进入函数前，load对象到函数调用的栈
# getrefcount()总是比实际存在的引用数多1
# 在函数中，有形参str保存了对实参a的引用

# 离开函数，引用计数-2
# 离开函数作用域，形参str销毁
# 函数后pop_top弹出函数栈暂存的引用

def ref_method(str):
    print(sys.getrefcount(str))
    print("我调用了{}".format(str))
    print('方法执行完了')

def ref_count():
    # 引用计数增加的场景
    print('测试引用计数增加')
    a = 'A'                    # 被创建并赋值给某个变量
    print(sys.getrefcount(a))  # 79
    b = a                      # 相互引用
    print(sys.getrefcount(a))  # 80
    ref_method(a)              # 作为参数 82
    print(sys.getrefcount(a))  # 恢复 80
    c = [1, a, 'abc']          # 放到某个容器对象中
    print(sys.getrefcount(a))  # 81

    # 引用计数减少的场景
    print('测试引用计数减少')
    del b                      # 引用变量被销毁
    print(sys.getrefcount(a))  # 80
    c.remove(a)                # 被从容器对象中移除
    print(sys.getrefcount(a))  # 79
    del c                      # 整个容器销毁（但a已经移除）
    print(sys.getrefcount(a))  # 79
    a = 783                    # 引用被赋值给其他对象
    print(sys.getrefcount(a))  # 4

if __name__ == '__main__':
    ref_count()
```

### 类定义——属性 和 方法函数
定义一个类  
class：

### Python对象的私有变量和内置方法
默认情况下，Python中的成员函数和成员变量都是公开的(public)。  
在python中定义私有变量只需要在变量名或函数名前加上 一个（保护）或两个（伪私有）下划线，那么这个函数或变量就是(伪)私有的了。  
公有变量可以直接访问，私有变量不可以被直接访问。  
单下划线为约定俗成的内部变量名。（在import操作中会不被包含）  
双下划线为对外访问隐藏名字，在继承操作中不被继承。  
伪私有变量可以通过 实例. \_类名\_\_变量名格式来强制访问。  
Python通过一个非常简单的机制完成了一个伪私有化功能，这个机制名叫名称转写(name mangling)：以双下划线开头，并以最多一个下划线结尾的标识符，例如__X，会被转写为_classname__X，其中classname为类名。这个机制实现起来非常简单，而且很大程度避免了调用者的误访问，但并不能像Java的private限定符那样完全杜绝外部的访问。

```python
class MyClass:
    __private = "私有" # Python会自动将 __private 重命名为 _MyClass__private
    
    def __init__(self):
        self.__instance_private = "实例私有"
        self._protected = "受保护属性"  # 单下划线，约定为内部使用
    def get_private(self):
        return self.__instance_private # 类内部可以访问

obj = MyClass()
print(dir(obj))  # 可以看到 '_MyClass__private', '_MyClass__instance_private'
print(obj._protected) # 受保护属性可以访问（但不推荐）
# print(obj.__instance_private) #  'MyClass' object has no attribute '__instance_private'
print(obj.get_private())
print(obj._MyClass__instance_private) # 通过修饰名强制访问私有变量
```

```python
# 序列对象常用的一些内置方法：
# seq.__iter__() # 遍历某个序列 iter(seq)
# seq.__next__() # 从迭代器中获取下一个值 next(seq)
# seq.__reversed__() # 按逆序创建一个迭代器 reversed(seq)
# seq.__len__() # 序列的长度 len(seq)
# seq.__contains__(x) # 了解某序列是否包含特定的值 x in seq
# x.__getitem__(key) # 通过键来获取值 x[key]
# x.__setitem__(key,value) # 通过键来设置值 x[key]=value
# x.__delitem__(key) # 删除一个键值对 del x[key]
# x.__missing__(nonexistent_key) # 为缺失键提供默认值 x[nonexistent_key]

# 定义一个迭代器类
class Foo:
    
    step = 2
    
    def __init__(self, n):
        self.n = n
        
    def __iter__(self):
        Foo.step = 3
        return self
    
    def __next__(self):
        self.n += Foo.step
        if self.n > 16:
            raise StopIteration
        return self.n

f1 = Foo(5)
print(next(f1))
print(next(f1))
for i in f1:
    print(i)
```

```python
# 可迭代的范围类（模拟range）
class MyRange:
    """模拟内置range的可迭代类"""
    def __init__(self, start, stop=None, step=1):
        if stop is None:
            self.start = 0
            self.stop = start
        else:
            self.start = start
            self.stop = stop
        self.step = step
        self.current = self.start
    
    def __iter__(self):
        self.current = self.start  # 重置迭代器
        return self
    
    def __next__(self):
        if (self.step > 0 and self.current >= self.stop) or \
           (self.step < 0 and self.current <= self.stop):
            raise StopIteration
        value = self.current
        self.current += self.step
        return value
    
    def __len__(self):
        """计算长度，如len(range(1,10,2))"""
        return max(0, (self.stop - self.start + self.step - 1) // self.step)
    
    def __reversed__(self):
        """支持逆序迭代"""
        stop = self.start - self.step
        start = self.stop - self.step
        return MyRange(start, stop, -self.step)
    
    def __contains__(self, item):
        """支持in操作符"""
        if (item - self.start) % self.step == 0:
            return self.start <= item < self.stop if self.step > 0 else \
                   self.start >= item > self.stop
        return False
    
    def __getitem__(self, key):
        """支持切片和索引"""
        if isinstance(key, slice):
            start = key.start or 0
            stop = key.stop or len(self)
            step = key.step or 1
            return MyRange(self.start + start * self.step,
                          self.start + stop * self.step,
                          self.step * step)
        else:
            if key < 0:
                key += len(self)
            if key < 0 or key >= len(self):
                raise IndexError("索引超出范围")
            return self.start + key * self.step

# 测试
r = MyRange(1, 10, 2)
print("原始:", list(r))
print("长度:", len(r))
print("包含5?", 5 in r)
print("包含6?", 6 in r)
print("逆序:", list(reversed(r)))
print("索引:", r[2])
print("切片:", list(r[1:4]))
```

```python
# 字典子类（带默认值和历史记录）
class HistoryDict(dict):
    """记录键访问历史的字典"""
    
    def __init__(self, default_factory=None, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.default_factory = default_factory
        self.history = {}  # 记录每个键的访问次数
    
    def __missing__(self, key):
        """处理缺失键"""
        if self.default_factory is None:
            raise KeyError(key)
        value = self.default_factory()
        self[key] = value
        return value
    
    def __getitem__(self, key):
        """记录访问历史"""
        if key not in self.history:
            self.history[key] = 0
        self.history[key] += 1
        return super().__getitem__(key)
    
    def __setitem__(self, key, value):
        """设置值时重置历史"""
        if key in self.history:
            self.history[key] = 0
        super().__setitem__(key, value)
    
    def __delitem__(self, key):
        """删除键和历史"""
        if key in self.history:
            del self.history[key]
        super().__delitem__(key)
    
    def get_history(self, key):
        """获取键的访问次数"""
        return self.history.get(key, 0)
    
    def most_accessed(self, n=5):
        """获取访问最频繁的n个键"""
        return sorted(self.history.items(), key=lambda x: x[1], reverse=True)[:n]

# 测试
hd = HistoryDict(list)  # 默认值为空列表
hd['a'].append(1)        # 自动创建键'a'，值为[]
print("hd['a']:", hd['a'])  # [1]，访问次数+1
hd['a'].append(2)        # 访问次数+1
hd['b'] = "直接赋值"      # 设置新键
print("访问a:", hd.get_history('a'))  # 2
print("访问b:", hd.get_history('b'))  # 0
print("最常访问:", hd.most_accessed())  # [('a', 2)]
```

```python
# 禁止添加新属性的类
def no_new_attributes(wrapped_setattr):
    def __setattr__(self, name, value):
        if hasattr(self, name):  # 已有属性，允许修改
            wrapped_setattr(self, name, value)
        else:  # 新属性，禁止添加
            raise AttributeError(f"不能添加新属性 {name}")
    return __setattr__

class NoNewAttrs:
    __setattr__ = no_new_attributes(object.__setattr__)

class Person(NoNewAttrs):
    name = '' # 已有属性
    def __init__(self, name):
        self.name = name
    def __repr__(self):
        return 'Preson(%r)' % (self.name)

p = Person("小明")
p.name = "小红"  # 修改已有属性，允许
print(p)
# p.age = 20     # AttributeError: 不能添加新属性 age
```

```python
# 树
class BinaryTree:
    def __init__(self, rootObj):
        self.key = rootObj        # 节点值
        self.leftChild = None     # 左子节点
        self.rightChild = None    # 右子节点
    
    def insertLeft(self, newNode):
        """插入左子节点"""
        if self.leftChild is None:
            self.leftChild = BinaryTree(newNode)  # 直接创建
        else:
            t = BinaryTree(newNode)          # 新节点
            t.leftChild = self.leftChild     # 原左子节点变成新节点的左子节点
            self.leftChild = t               # 新节点成为当前节点的左子节点
    
    def insertRight(self, newNode):  # 同上，插入右子节点
        if self.rightChild is None:
            self.rightChild = BinaryTree(newNode)
        else:
            t = BinaryTree(newNode)
            t.rightChild = self.rightChild
            self.rightChild = t
    
    def getLeftChild(self):   # 获取左子节点
        return self.leftChild
    
    def getRightChild(self):  # 获取右子节点
        return self.rightChild
    
    def setRootVal(self, obj):  # 设置节点值
        self.key = obj
    
    def getRootVal(self):       # 获取节点值
        return self.key
    
r = BinaryTree('a')
print(r.getRootVal())
print(r.getLeftChild())
r.insertLeft('b')
print(r.getLeftChild())
print(r.getLeftChild().getRootVal())
r.insertRight('c')
print(r.getRightChild())
print(r.getRightChild().getRootVal())
r.getRightChild().setRootVal('hello')
print(r.getRightChild().getRootVal())
```

### 类的继承
BaseClassName（示例中的基类名）必须与派生类定义在一个作用域内（使用import即将其放入同一作用域内）。  
派生类的定义同样可以使用表达式。  
创建一个新的类实例，方法引用按如下规则解析：搜索对应的类属性，必要时沿基类链逐级搜索，如果找到了函数对象这个方法就是合法的。

```python
# 第一种：直接继承（基类在当前作用域）
# BaseClassName 必须在当前作用域可以直接访问
# 即在当前文件中定义的或import进来的(from modname import BaseClassName)
# class DerivedClassName(BaseClassName):
#     # 类的内容
#     ……
# 第二种：通过模块名继承（基类在其他模块）
# 不先导入，直接在继承时指定完整路径
# modname.BaseClassName：通过模块名.类名指定基类
# class DerivedClassName(modname.BaseClassName):
#     # 类的内容
#     ……

class Person(object):   # 父类（基类）
    def talk(self):     # 父类中的方法
        print("person is talking...")

class Chinese(Person):  # 子类（派生类），继承Person
    def walk(self):     # 在子类中定义其自身的方法
        print('Chinese person is walking...')

c = Chinese()
c.talk()  # 使用父类方法
c.walk()  # 使用自己的方法
```

```python
# 静态方法：不会被重新创建，直接按名引用
class Book:
    def __init__(self, title):
        self.title = title
    
    # 普通实例方法：需要实例，第一个参数是 self
    def normal_method(self):
        return self.title  # 需要self访问实例属性
    
    # 类方法（@classmethod）：第一个参数是 cls
    @classmethod
    def class_method_create(cls, title):  # cls代表类本身
        book = cls(title=title)  # 使用cls创建实例
        return book
    
    # 静态方法（@staticmethod）：没有特殊参数
    @staticmethod
    def static_method_create(title):
        book = Book(title)  # 硬编码类名
        return book


book1 = Book("普通方式")
book2 = Book.class_method_create("类方法方式")
book3 = Book.static_method_create("静态方法方式")
print(book1.title)
print(book2.title)
print(book3.title)
```

```python
# 静态方法设定为恒定的当前运行环境，类方法的运算环境可以随着继承关系而进化。
class Foo:
    X = 1
    Y = 14
    
    @staticmethod
    def static_method():  
        return Foo.X + Foo.Y  # 硬编码Foo，永远使用Foo的属性
    # 静态方法由于不使用相对引用来标定参数，因此不会随着继承到新环境而改变运算逻辑。
    
    @classmethod
    def class_method(cls):
        return cls.X + cls.Y  # 使用cls，动态查找当前类的属性
    # 类方法由cls参数自动带入类的环境引用，因此会随着继承到新环境而改变运算逻辑。

class Son(Foo):
    X = 3  # 重写类属性
    Y = 5

# 测试
print("Foo.static_method():", Foo.static_method())  # 15 (1+14)
print("Foo.class_method():", Foo.class_method())    # 15 (1+14)

print("Son.static_method():", Son.static_method())  # 15 (仍然1+14，硬编码了Foo)
print("Son.class_method():", Son.class_method())    # 8 (3+5，使用Son的属性)
```

# Python设计模式
设计模式是面向对象方法里的一种解决方案的抽象。目的是把一些常见的应用抽象为一种类设计模式，在具体实现中只要套用或稍作修改，就能完成逻辑清晰的类实现方案。  
通过对设计模式的学习也可以达到对类体系的深入理解。

```python
# 单例模式 — 所有生成实例都指向同一个对象，确保一个类只有一个实例
# 方法1：重载 __new__
class Singleton:
    _instance = None
    def __init__(self):
        print("Do something.")
    def __new__(cls, *args, **kwargs): # 重载__new__()方法
        if not cls._instance:
            cls._instance = super().__new__(cls)
        return cls._instance
# 另一种实现单例模式的方案
# class Singleton:
#     def __init__(self):
#         print("Do something.")
#     def __new__(cls, *args, **kwargs):
#         if not hasattr(Singleton,"_instance"): # 在cls空间动态生成一个内部属性
#             Singleton._instance = super().__new__(cls)
#         return Singleton._instance

obj1 = Singleton()
obj2 = Singleton()
print(obj1 is obj2)  # True，同一个对象

# 方法2：装饰器
instances = {} # 一个全局的 类-实例对 的记忆buffer

def singleton(cls): # 接收一个类
    def get_instance(*args, **kwargs):
        cls_name = cls.__name__
        if cls_name not in instances: # 过去没有创建实例
            print("第一次创建实例")
            instance = cls(*args, **kwargs) # 创建实例
            instances[cls_name] = instance # 加入记忆buffer
        else:
            print("已经创建过实例了")
        return instances[cls_name] # 返回实例引用
        
    return get_instance

@singleton
class User:
    def __init__(self,name):
        self.name = name

xiaowang = User("xiao wang")
xiaoming = User("xiao ming")
print(xiaoming is xiaowang)
```

```python
# 工厂模式简介：用来实现参数化定制类
# 工厂模式的本质上是用类定制类，然后到具体实例。
# 工厂模式的基础是共性抽象，是把相关类的共性和个性化定制相融合的解决方案。
# 目的：创建对象时不暴露创建逻辑

class StandardFactory:
    @staticmethod
    def get_factory(factory):
        if factory == "dog":
            return DogFactory()
        elif factory == "cat":
            return CatFactory()
        else:
            raise ValueError("Unknown Factory.")

class DogFactory:
    def get_pet(self):
        return Dog()

class CatFactory:
    def get_pet(self):
        return Cat()

# 抽象类
from abc import ABC, abstractmethod
# 抽象类可以通过MyIterable方法来查询所有的派生子类
class Animal(ABC):  # 抽象基类
    @abstractmethod
    def speak(self):  # 抽象方法，强制子类必须实现
        pass
    
    def eat(self):  # 普通方法，不能创建实例，但可以被继承
        print("吃东西")

class Dog(Animal):
    def speak(self):  # 必须实现抽象方法
        print("汪汪")

class Cat(Animal):
    def speak(self):
        print("喵喵")

# animal = Animal()  # 不能实例化抽象类
factory = StandardFactory.get_factory('cat')
cat = factory.get_pet()
cat.speak()
factory = StandardFactory.get_factory('dog')
dog = factory.get_pet()
dog.speak()
```

```python
# 观察者模式（发布-订阅模式？）
# 多使用一种“注册—通知—撤销注册”的形式。
# （Observer）将自己注册到被观察对象（Subject）中，被观察对象将观察者存放在一个容器（Container）里。
# 被观察对象发生了某种变化，从容器中得到所有注册过的观察者，将变化通知观察者。

class Observer:
    """观察者"""
    def __init__(self, name):
        self.name = name
    
    def update(self, message):
        print(f"{self.name}收到：{message}")

class Subject:
    """被观察的主题"""
    def __init__(self):
        self.observers = []
    
    def add_observer(self, observer):
        self.observers.append(observer)
    
    def remove_observer(self, observer):
        self.observers.remove(observer)
    
    def notify(self, message):
        for observer in self.observers:
            observer.update(message)

# 使用
weather = Subject()

xiaoming = Observer("小明")
xiaohong = Observer("小红")

weather.add_observer(xiaoming)
weather.add_observer(xiaohong)

weather.notify("今天下雨！")

weather.remove_observer(xiaohong)
weather.notify("天晴了！")
```

### 正则表达式
可以用来判断某个字符串是否符合某种模式，比如判断某个字符串是不是邮箱地址，网址，电话号码，身份证号  
可以用来在文本中寻找并抽取符合某种模式的字符串  
正则表达式规定了一种模式，是一个字符串，如“abc”,“a.?p.*c”

正则表达式字符：  
一般字符  代表自身。  
.  匹配任意除换行符\n外的字符。  
\  转义字符，使后一个字符改变原来的意思。  
[…]  字符集。对应的位置可以是字符集中任意字符。  
[^abc]  第一个字符是^表示取反，即匹配abc之外的任意一个字符。  
预定义字符，可以在字符集中使用：  
\d  数字[0-9]  
\D  非数字[^\d]  
\s  空白字符[<空格>\n\t\r]  
\S  非空白字符  
\w  单词字符[A-Za-z0-9]  
\W  非单词字符[^\w]  
数量词字符：  
\*  匹配前一个字符0或无数次  
\+  匹配前一个字符1次或无数次  
？ 匹配前一个字符0次或1次  
{m}  匹配前一个字符m次  
{m, n}  匹配前一个字符m至n次，可省略  
[\dabc]+ 长度不为0的由数字或a、b、c构成的字符串  
字符边界:  
\A  与字符串开始的位置匹配，不消耗任何字符。  
\Z  与字符串结束的位置匹配，不消耗任何字符。  
^  与字符串开始的位置匹配，不消耗任何字符。在多行模式中，匹配每一行开头。  
$  与字符串结束的位置匹配，不消耗任何字符。在多行模式中，匹配每一行末尾。  
\b  匹配一个单词的开始处和结束处，不消耗任何字符。  
\B  和\b相反，不允许是单词的开始处和结束处。

### Python与正则表达式——import re

```python
# re.match(pattern, string, flags=0) 
# 从字符串的起始位置一个模式pattern。
# flags是标志位，用于控制模式串的匹配方式，如是否区分大小写，多行匹配等，匹配成功则返回一个匹配对象，失败返回None。
# re.I 使匹配不区分大小写。
# re.M 跨多行匹配，影响^和$。
# re.S 使.匹配所有字符。
import re

m1 = re.match("ab*c", "abbcde")
print('m1: ', m1 != None)
m2 = re.match("ab+c", "ac")
print('m1: ', m2 != None)
m3 = re.match("a.?bc.*", "abcd")
print('m1: ', m3 != None)
m4 = re.match("a.?bc.*", "aBcd")
print('m1: ', m4 != None)
m5 = re.match("a.?bc.*", "aBcd", re.I) # 不区分大小写
print('m1: ', m5 != None)
```

```python
# re.search(pattern, string[, flags]) 
# 查找字符串中可以匹配成功的子串。
# 匹配成功则返回一个匹配对象，失败返回None。
import re

print('1: ', re.search("a.+bc*$", "mnaedwbc"))
print('2: ', re.search("a.+bc*$", "mnaedwbcd"))  
print('3: ', re.search("^a.+bc*", "mnaedwbcd"))
print('4: ', re.search("^a.+bc*", "adewbcd"))
```

```python
# re.findall(pattern, string[, flags]) 
# 查找字符串中所有和模式匹配的子串放入列表，一个子串都找不到则返回[]
import re

# 示例1：查找所有数字
text = "我的电话是12345，办公室电话是67890"
result = re.findall(r"\d+", text)
print(result)  # 输出：['12345', '67890']

# 示例2：查找所有邮箱
emails = "联系邮箱: test@test.com, admin@test.com, 或者support@test.com"
result = re.findall(r"\w+@\w+\.\w+", emails)
print(result)  # 输出：['test@test.com', 'admin@test.com', 'support@test.com']

# 示例3：找不到返回空列表
result = re.findall(r"\d+", "没有数字的一句话")
print(result)  # 输出：[]
```

```python
# 字符边界
import re

# 1. \A：匹配字符串开头（忽略多行模式）
test1 = r"\Ahow are"
print('test1-1: ', re.search(test1, "hhow are you"))
print('test1-2: ', re.search(test1, "how are you").group())

# 2. \Z：匹配字符串结尾（忽略多行模式）
test2 = r"are you\Z"
print('test2-1: ', re.search(test2, "how are you?"))
print('test2-2: ', re.search(test2, "how are you").group())

# 3. ^：匹配行首（多行模式re.M）
test3 = r"^how are"
print('test3-1: ', re.findall(test3, "how are you\nhow are you", re.M))

# 4. $：匹配行尾（多行模式re.M）
test4 = r"are you$"
print('test4-1: ', re.search(test4, "how are you\nhow are you?", re.M).group())

# 5. \b：单词边界
test5 = r"\bA.*B\b C"
print('test5-1: ', re.search(test5, "Abb$BD CD"))
print('test5-2: ', re.search(test5, "test Abb$B CD").group())

# 6. \B：非单词边界
test6 = r"\BA.*B\B\w C"
print('test6-1: ', re.search(test6, "test Aab$B CD"))
print('test6-2: ', re.search(test6, "testAab$BE CD").group())
```

```python
# 正则表达式中的“分组”
# 括号中的表达式是一个分组，多个分组按左括号从左到右从1开始编号。
import re

pattern = "(((ab*)c)d)e"
r = re.match(pattern, "abbcdefg")

print('group:', r.groups())      # 所有分组（元组）
print('index:', r.lastindex)     # 最后一个分组的索引
print('group 0:', r.group(0))    # 整个匹配
print('group 1:', r.group(1))    # 第1个括号
print('group 2:', r.group(2))    # 第2个括号  
print('group 3:', r.group(3))    # 第3个括号
```

```python
# re.finditer(pattern, string[, flags]) 
# 在字符串中找到正则表达式所匹配的所有子串，并作为一个迭代器返回
import re

s = '123[45]67<890>a[bc]ba<098>'
m = r'\[(\d+)\]|<(\d+)>'  # 匹配[数字] 或 <数字>

for x in re.finditer(m, s):
    print(x.group())  # 输出整个匹配


p = r"(((ab*)+c|12+3)d)e"  # 匹配 "ab...c)d)e" 或 "1223)d)e"

for x in re.finditer(p, 'ababcdef12def1223def'):
    print(x.group()) # group等价于group(0)
    print(x.groups()) # groups返回一个元组，元素一次是1-3号分组所匹配的字符串


m = re.finditer(r'([a-z]+)','one1;two32 three3,four4') # 匹配由a-z组成的串无数次，并作为迭代器返回
print(next(m)) # <re.Match object; span=(0, 3), match='one'>
print(next(m)) # <re.Match object; span=(5, 8), match='two'>
print(next(m)) # <re.Match object; span=(11, 16), match='three'>
print(next(m)) # <re.Match object; span=(18, 22), match='four'>

m = re.finditer(r'([a-z]+)([0-9]+)','one1;two32 three3,four4') # 匹配由(a-z)+(0-9)组成的串无数次，并作为迭代器返回,模式1为(a-z)，模式2为(0-9)
print(next(m)) # <re.Match object; span=(0, 4), match='one1'>
print(next(m).group())  # 整个模式 two32
print(next(m).group(1)) # 子模式1 three3中的three
print(next(m).group(2)) # 子模式2 four4中的4
```

```python
# 数量词的贪婪模式与非贪婪模式
import re

# 贪婪模式：默认情况下，量词+,*,?,{m,n}会匹配尽可能长的字符串
print("【贪婪模式】")
pattern_greedy = "<t>.*</t>"  # .* 会匹配到最后一个可能的</t>
string = "<t>abcd</t><t>abcde</t>"
test1 = re.match(pattern_greedy, string)
print('test1: ', test1.group())

print("【非贪婪模式】")
# 非贪婪模式：在量词后面加'?'，匹配尽可能短的字符串
pattern_non_greedy = "<t>.*?</t>"  # .*? 会匹配到第一个</t>就停止
test2 = re.match(pattern_non_greedy, string)
print('test2: ', test2.group())
```

```python
# 正则表达式自定义命名分组(?P)
# 一个正则表达式可以有多个自定义名称的分组，可以通过分组名称提取匹配的字符串，每一个分组的格式是：
# (?P<自定义分组名称>正则字符串)
import re

def addOne(matched):
    """将匹配到的数字加1"""
    value = int(matched.group('int'))  # 通过分组名称提取
    return str(value + 1)

string = 'A123B4CD56EF789G'
result = re.sub('(?P<int>\d+)', addOne, string)
print('原始字符串:', string)
print('替换后:', result)
```

# 进程、线程与协程

### 并行/并发编程
* 并行：如多核CPU不同核上跑的两个进程。两个计算流在时间上重叠。
* 并发：如单核CPU上跑的两个进程。两个计算流在时间上交替执行。给我们带来宏观上两个进程同时运行的假象。

程序执行时的私有环境：  
程序计数器：控制读到哪一句  
通用目的寄存器、浮点寄存器、条件码寄存器：存某些变量的值、中间计算结果、栈指针、子程序返回值……  

计算机如何执行程序？  
内核在负责资源的调度。每个进程有独立的逻辑控制流、私有的虚拟地址空间。   
维护一个进程需要：程序计数器、通用目的寄存器、浮点寄存器、状态寄存器；用户栈、内核栈、内核数据结构（用来映射虚拟地址的页表、当前打开文件信息的文件表）；进程的上下文。  
具体执行中是通过进程头与CPU的寄存器组配合保留-回复进程的私有运行环境。  

内核：分时调度和共享资源  
内核负责进程的挂起和唤醒，在进程执行期间进行上下文的切换。  
进程（process）：虚地址空间独立运行。其资源被内核保护起来。  
线程（thread）：调度执行的最小单元。每个线程运行在单一进程中。一个进程中可以有多个线程。


### Python中的并发处理
* 多进程  
程序之间不共享内存，使用虚拟地址映射保证资源独立。  
优点：一个进程挂了不影响别的进程。  
缺点：切换上下文效率低，通信和信息共享不太方便。  
* 多线程  
优点：上下文切换效率比多进程高，线程之间信息共享和通信方便。  
缺点：一个线程挂了会使整个进程挂掉。操作全局变量需要锁机制。  

不同的进程、线程，之间总是并发/并行的。如果运行在多核CPU不同的核上，是可以并行的。

### Python多进程编程
多进程模式：Process对象  
p.start()：启动进程,并调用子进程中的p.run()。  
p.run()：进程启动时运行的方法，正是他去调用target指定的函数。  
p.terminate()：强制终止进程p，不会进行任何清理操作,如果p创建了子进程,该子进程就成了僵尸进程。使用该方法需要特别小心这种情况，如果p还保存了一个锁那么也将不会被释放，进而导致死锁。  
p.is_alive()：如果p仍然运行，返回True。  
p.join ([timeout])：主线程等待p终止（是主线程处于等待的状态，而p是处于运行状态），timeout是可选的超时时间。需要强调的是，p.join只能join start开启的进程，而不能join run开启的进程。  
对主进程来说，运行完毕指的是主进程代码运行完毕。  
对主线程来说，运行完毕指的是主线程所在的进程内所有非守护线程统统运行完毕，主线程才算运行完毕。  

注意：由于Jupyter notebook自身原因，只能跟踪主进程，没法跟踪子进程，所以在这里使用multiprocessing，需要将脚本另存为py文件，然后再运行。  
Jupyter Notebook里直接运行asyncio.run(main())会报错RuntimeError: asyncio.run() cannot be called from a running event loop，是因为ipykernel本身运行在一个event loop上，而asyncio是不支持嵌套的，这也是notebook里能直接调用await main()的原因。
在.py文件中await必须出现在async function里面。

```python
%%writefile test.py
from multiprocessing import Process
import os
import time

def hello(i):
    print("son process id {} - for task {}".format(os.getpid(), i),flush=True)
    time.sleep(0.1)

if __name__ == '__main__':
    print("current father process {}".format(os.getpid()),flush=True)
    start = time.time()
    p1 = Process(target=hello, args=(1,))
    p2 = Process(target=hello, args=(2,))
    p1.start()
    p2.start()
    p1.join() # wait until it finishes
    p2.join()
    end = time.time()
    print("Totally take {} seconds".format((end-start)))
```

### 多进程池（并发计算资源共享）
进程池:  
定义一个池子, 在里面放上固定数量的进程 , 有任务要处理的时候就会拿一个池中的进程来处理任务, 等到处理完毕, 进程并不关闭而是放回进程池中继续等待任务。  
如果很多任务需要执行, 池中的进程数不够, 任务会就要等待, 拿到空闲的进程才能继续执行。  

multiprocessing.Pool模块：  
Pool ( [numprocess [ ,initializer [, initargs] ] ] ) : 创建进程池  
numprocess：要创建的进程数, 如果省略, 将默认使用os.cpu_count() 的值  
initializer：是每个工作进程启动时要执行的可调用对象, 默认为None  
initargs：传给initializer的参数组  

多进程任务调用方式：  
p.apply(func [ ,args [ ,kwargs]])：在一个池工作进程中执行func(*args,\*\*kwargs), 然后返回结果（同步调用，阻塞主进程）  
p.apply_async(func [ ,args [ ,kwargs]])：在一个池工作进程中执行func(*args,\*\*kwargs), 然后返回结果（异步调用）  
p.close()：关闭进程池, 防止进一步操作. 如果所有操作持续挂起, 他们将在工作进程终止前完成  
p.join()：等待所有工作进程退出. 此方法只能在close()或terminate()之后调用  

```python
%%writefile test.py
from multiprocessing import Process, Pool
import os
import time

def hello(i):
    print("son process id {} - for task {}".format(os.getpid(), i),flush=True)
    time.sleep(0.1)

if __name__ == '__main__':
    print("current father process {}".format(os.getpid()),flush=True)
    start = time.time()
    p = Pool(4)
    for i in range(5): #  task 0，1，2，3是立刻执行的，而task 4要等待前面某个task完成后才执行
        p.apply_async(hello,args=(i,))
    # apply_async是异步的，子进程执行的同时，主进程继续向下执行。
    print("Waiting for all subprocesses done...",flush=True)
    # 所以“Waiting for all subprocesses done...”先打印出来，“All subprocesses done.”最后打印。
    p.close() # no longer receive new process
    p.join() # wait until all processes in the pool finishes
    print("All subprocesses done.")
    end = time.time()
    print("Totally take {} seconds".format((end-start)))
```

### 进程间通讯
两个进程传递消息，可以使用Pipe() 。多个进程间共享消息或数据可以采用队列Queue（允许多个生产者和消费者）。  
尽量避免自己设计资源锁模式进行同步。  
multiprocessing使用queue.Empty和queue.Full异常。  
Queue 有两个方法，get 和 put （可以设定阻塞或非阻塞）  
q.put(item)：将item放入队列中, 如果当前队列已满, 就会阻塞, 直到有数据从管道中取出  
q.put_nowait(item)：将item放入队列中, 如果当前队列已满, 不会阻塞, 但是会报错  
q.get()：返回放入队列中的一项数据, 取出的数据将是先放进去的数据, 若当前队列为空, 就会阻塞, 直到放入数据进来  
q.get_nowait() 返回放入队列中的一项数据, 同样是取先放进队列中的数据, 若当前队列为空, 不会阻塞, 但是会报错  

```python
%%writefile test.py
# 进程的主从通讯
# Pipe() 返回一个由管道连接的连接对象，默认情况下是双工（双向）
from multiprocessing import Process, Pipe

def f(conn):
    conn.send([520, None, 'hello'])
    print(conn.recv())
    conn.close()

if __name__ == '__main__':
    parent_conn, child_conn = Pipe()
    p = Process(target=f,args=(child_conn,))
    p.start()
    print(parent_conn.recv(),flush=True)
    parent_conn.send('Hi')
    p.join()
```

### 线程
由于进程之间不共享内存，进程的切换-协同效率低。  
线程（thread）：调度执行的最小单元。每个线程运行在单一进程中。一个进程中可以有多个线程。  
线程上下文：程序计数器、通用目的寄存器、浮点寄存器、条件码，线程ID，栈，栈指针  
线程间可以共享进程空间的公有数据：进程打开的文件描述符、信号-锁处理器、进程的当前目录和进程用户ID与进程组ID。因此可以方便的进行通讯。  

```python
import threading
import time

def hello(i):
    time.sleep(0.000001)
    print("thread id:{} for task{}\n".format(threading.current_thread().name, i),end='',flush=True)
    time.sleep(0.000001)

if __name__ == '__main__':
    start = time.time()
    t1 = threading.Thread(target=hello,args=(1,))
    t2 = threading.Thread(target=hello,args=(2,))
    t1.start()
    t2.start()
    t1.join()
    t2.join()
    end = time.time()
    print("Take {}s".format((end-start)))
```

### 线程锁机制：  
进程间可以共享全局变量。多个逻辑控制流同时读写共享的资源时，需要加入锁（Lock）机制。  
例：100个线程，每个都对一个全局变量cnt不加锁操作cnt+=1。执行完后结果≤100  
加锁：  
需要读写某个共享变量时，读写前上锁，读写后释放锁。  
如果要读写的共享变量上了锁，等待锁释放后再上锁读写。  

全局解释器锁：一个进程任一时刻仅有一个线程在执行。  
在CPython中由于全局解释器锁（GIL）的存在，多核CPU并不能为它显著提高效率。可以考虑选择没有GIL的Python解释器（如JPython）。

```python
%%writefile test.py
from multiprocessing import Process, Lock
import time

def f(i, l):
    # 未加锁
    # l.acquire() # 加锁
    try:
        print("hello", i, flush=True)
        time.sleep(0.1)
        print(i, "do something", flush=True)
    finally:
        # l.release() # 保证锁释放
        pass

if __name__ == '__main__':
    lock = Lock()
    
    for num in range(5):
        Process(target=f,args=(num,lock)).start()
```

### 资源管理器：独占同时防止意外死锁
multiprocessing.Manager() 提供了一个服务器进程，允许多个进程通过代理对象（Proxy）共享数据结构（如 list、dict、Namespace、Queue、Lock 等），并且支持跨网络共享。这种方式比共享内存更灵活，但性能略低。  
这里 manager.list() 和 manager.dict() 返回的是进程安全的代理对象，可在不同进程中直接修改并同步。  

关键特性与注意事项：  
跨进程共享：Manager 创建的对象可被多个进程访问，甚至可通过网络共享。  
线程安全：内部已实现同步机制，避免数据竞争。  
性能：由于通过进程间通信（IPC）实现，速度比 Value/Array 共享内存慢。  
上下文管理：推荐 with Manager() as m: 自动启动与关闭管理进程，避免资源泄露。  
嵌套支持：共享容器可包含其他共享对象（如 manager.list([manager.dict()])）。  

```python
%%writefile test.py
from multiprocessing import Process, Manager
import time

def f(i, d, l):
    print("task {}".format(i), flush=True)
    d[1] = 'task {}'.format(i)
    d['2'] = 2 + i
    d[0.25] = i
    time.sleep(0.1)
    l.reverse()

if __name__ == '__main__':
    with Manager() as manager: # 类似上下文管理器，先加再解
        d = manager.dict() # Manager 共享对象
        l = manager.list(range(10))
        
        p1 = Process(target=f,args=(1,d,l))
        p2 = Process(target=f,args=(2,d,l))
        p1.start()
        p2.start()
        p1.join()
        p2.join()
        
        print(d)
        print(l)
    # 退出时自动关闭 manager（manager.shutdown()），清理资源
```

### 协程
本质上是一个可被异步唤醒的函数  
存在自阻塞操作（语句）的被动服务函数  
存在发出调用操作并等待调用返回才继续执行的主动客户方  
协程的调度由用户程序（而非系统内核）自己来控制  

```python
# 基础协程示例：带接收功能的生成器
def fib(n):
    index = 0
    a = 0
    b = 1
    
    while index < n:
        receive = yield b          # yield b 产出值，同时可以接收外部发送的值
        print(f"fib receive {receive}")
        a, b = b, a + b
        index += 1

# 使用示例
fib_gen = fib(20)
print(f"fib yield {fib_gen.send(None)}")  # 第一次必须send(None)或next(fib)，用于启动协程
print(f"fib yield {fib_gen.send(1)}")     # send(1) 给receive赋值1
print(f"fib yield {fib_gen.send(3)}")
print(f"fib yield {fib_gen.send(5)}")
```

```python
# 带条件的协程：根据接收值改变行为
def gen_cal():
    x = 1
    y = 1
    exp = None
    
    while x < 256:
        if exp == None:           # 第一次运行，没有接收值
            x, y = y, x + y
            exp = yield y          # 产出y，并接收值赋给exp
        else:
            exp = yield (eval(exp)) # 接收表达式，计算后产出结果

# 测试
gc = gen_cal()
print(next(gc))
print(next(gc))
print(gc.send('23+9/3.0')) # 计算表达式
print(next(gc))
```

```python
# 任务轮转调度，用协程实现
def task1():
    timeN = 12
    dur = 6
    
    while timeN > 0:
        timeN -= dur
        print(f'Task1 need: {timeN}')
        yield timeN          # 返回剩余时间并暂停
    
    print('Task 1 Finished')

def task2():
    timeN = 11
    dur = 3
    
    while timeN > 0:
        timeN -= dur
        print(f'Task2 need: {timeN}')
        yield               # 只交出控制权，不返回值
    
    print('Task 2 Finished')

def RoundRobin(*tasks):
    tls = list(tasks)
    
    while len(tls) > 0:
        for p in tls[:]:     # 使用切片复制，避免修改时出错
            try:
                next(p)
            except StopIteration:
                tls.remove(p)
    
    print('All finished! 可以歇一会了')

# 运行
t1 = task1()
t2 = task2()
RoundRobin(t1, t2)
```

```python
# 高级调度器：带优先级和包装类
waiting_list = []

class Handle(object):
    """任务包装类，可以设置优先级、时间片等"""
    
    def __init__(self, gen, pri=0.5):
        self.gen = gen
        self.timeSlice = 0
        self.timeNeed = 0
        self.pr = pri          # 优先级，越高时间片越大
    
    def call(self):
        try:
            if self.timeSlice == 0:
                # 首次调用，获取需要的执行时间
                self.timeNeed = next(self.gen)
                self.timeSlice = int(self.timeNeed * self.pr)
            else:
                # 非首次调用，发送时间片
                self.gen.send(self.timeSlice) # 运行，并设置下一个时间片长度
            
            waiting_list.append(self)  # 放回队列
            
        except StopIteration:
            print(f'{self.gen.__name__} finished')

def RoundRobin(*tasks):
    # 将所有任务包装后加入队列
    for task in tasks:
        waiting_list.append(Handle(task))
    
    while waiting_list:
        p = waiting_list.pop(0)   # 从队头取出
        p.call()                  # 执行
    
    print('All finished! 可以歇一会了')

# 定义任务函数
def task1():
    need = 12
    while need > 0:
        time_slice = yield need    # 接收时间片
        print(f'Task1 need: {need}',end=' , ')
        print(f'Time slice = {time_slice}')
        need -= time_slice
        print(f'Task1 remaining: {need}')

def task2():
    need = 11
    while need > 0:
        time_slice = yield need
        print(f'Task2 need: {need}',end=' , ')
        print(f'Time slice = {time_slice}')
        need -= time_slice
        print(f'Task2 remaining: {need}')

# 运行
t1 = task1()
t2 = task2()
RoundRobin(t1, t2)
```

```python
%%writefile test.py
# 顺序执行
import asyncio
from datetime import datetime

async def say_after(delay, what):
    await asyncio.sleep(delay)  # 模拟IO操作
    print(what)

async def main():
    print(f"started at {datetime.now().strftime('%H:%M:%S.%f')[:-4]}")
    
    await say_after(0.1, 'hello')  # 等待0.1秒
    await say_after(0.2, 'world')  # 等待0.2秒
    # 经过0.1+0.2=0.3s
    print(f"started at {datetime.now().strftime('%H:%M:%S.%f')[:-4]}")

asyncio.run(main())
```

```python
%%writefile test.py
# 并发执行：create_task()
import asyncio
from datetime import datetime

async def say_after(delay, what):
    await asyncio.sleep(delay)  # 模拟IO操作
    print(what)

async def main():
    # 创建任务（立即开始执行，不等待）
    task1 = asyncio.create_task(say_after(0.3, 'hello'))
    task2 = asyncio.create_task(say_after(0.2, 'world'))
    
    print(f"started at {datetime.now().strftime('%H:%M:%S.%f')[:-4]}")
    
    # 等待任务完成
    await task1
    await task2
    # 经过max(0.3,0.2)=0.3s
    print(f"started at {datetime.now().strftime('%H:%M:%S.%f')[:-4]}")

asyncio.run(main())
```

```python
%%writefile test.py
# 用事件循环队列实现生产者-消费者异步流程
import asyncio
import time

async def consumer(q):
    print('consumer starts.')
    while True:
        item = await q.get()  # 从队列获取数据（如果没有则等待）
        
        if item is None:      # None 是结束信号
            q.task_done()
            break
        else:
            await asyncio.sleep(0.1)  # 模拟消费耗时
            print('consume %d' % item)
            q.task_done()      # 通知队列一个任务已完成
    
    print('consumer ends.')

async def producer(q):
    print('producer starts.')
    
    for i in range(5):
        await asyncio.sleep(0.1)  # 模拟生产耗时
        print('produce %d' % i)
        await q.put(i)          # 放入队列
    
    await q.put(None)           # 发送结束信号
    await q.join()              # 等待所有消费完成
    print('producer ends.')

# 主程序
q = asyncio.Queue(maxsize=10)
t0 = time.time()
loop = asyncio.get_event_loop() # 获取当前事件循环
tasks = [producer(q), consumer(q)]
loop.run_until_complete(asyncio.wait(tasks))
# asyncio.wait(tasks) 创建一个新的协程，等待多个任务完成
# loop.run_until_complete() 启动事件循环，运行指定的协程直到它完成
loop.close() # 关闭事件循环，释放资源
print(time.time() - t0, "s")
```

# Python 网络编程简介
什么是计算机网络？
计算机网络：用通信设备将计算机连接起来，在计算机之间传输数据（信息）的系统。  
连网的计算机根据其提供的功能将之区分为客户机或服务器（C/S）。  
通信协议：计算机之间以及计算机与设备之间进行数据交换而遵守的规则、标准或约定。  
典型的协议：TCP/IP（在互联网上采用），IEEE802.3以太网协议（局域网），IEEE902.11（无线局域网，WIFI）  

### 开放系统互连模型OSI
* 物理层：利用传输介质为数据链路层提供物理连接，实现比特流的透明传输。
* 数据链路层：将比特组成数据帧，临近网络部件间的数据传送
* 网络层：源到目的之间基于分组的数据交换
* 传输层：提供端对端的透明数据传输服务
* 会话层：不同主机的进程间会话的组织和同步
* 表示层：为上层用户提供共同需要的数据或信息语法表示及转换
* 应用层：为用户提供服务，提供网络应用

### 客户端-服务器模型
一个应用由一个服务器进程和一个或多个客户端进程组成。  
服务器管理某种资源，并且通过操作这种资源来为它的客户端提供某种服务。  
1. 客户端向服务器发送一个请求。  
Web浏览器需要文件时，发送请求给Web服务器。  
2. 服务器收到请求后，解释它，并以适当的方式操作资源。  
Web服务器收到浏览器发出的请求后，读一个磁盘文件。  
3. 服务器给客户端发送一个响应，并等待下一个请求。  
Web服务器将文件发送回客户端。  
4. 客户端收到响应并处理它。  
Web浏览器收到来自服务器的一页后，在屏幕上显示此页。  

以买火车票为例：  
客户端：发出查询请求，如果有则购买一张票。  
服务端：维护余票情况，如果有余票则卖票给客户端，余票数量减一；没有则返回购买失败。  
B/S模式：Browser/Server，对C/S模式的改进。一部分事务逻辑在前端实现，但是主要事务逻辑在服务器端实现，和数据库端形成三层结构。建立在广域网之上，只要有网络、浏览器，可以随时随地进行业务处理。  

### 传输层协议
* TCP：传输控制协议，面向连接、可靠。适用于要求可靠传输的应用。  
面向连接：发送数据之前必须在两端建立连接。  
仅支持单播传输：只能进行点对点数据传输。  
面向字节流：在不保留报文边界的情况下以字节流的方式进行传输。  
可靠：对每个包赋予序号，来判断是否出现丢包、误码。  
* UDP：用户数据报协议，面向非连接、不可靠。适用于实时应用。  
面向非连接：发送数据不需要建立连接。  
支持单播、多播、广播。  
面向报文：对应用层的报文添加首部后直接向下层交付。  
不可靠：没有拥塞控制，不会调整发送速率。  

网络通讯实现的基本方案：服务器监听特定的端口：如8080；得到请求后建立数据链路（远程I/O strem）；启动阻塞式通讯（TCP）  

### IP地址
IPv4 – 32位，IPv6 – 128位  
IP地址分类：每个地址由两个固定长度的字段组成，网络号net-id标志主机所连接到的网络，主机号host-id标志该主机。  
127.0.0.1和0.0.0.0的区别：  
回环地址127.x.x.x：该范围内的任何地址都将环回到本地主机中，不会出现在任何网络中。主要用来做回环测试。  
0.0.0.0：任何地址，包括了环回地址。不管主机有多少个网口，多少个IP，如果监听本机的0.0.0.0上的端口，就等于监听机器上的所有IP端口。数据报的目的地址只要是机器上的一个IP地址，就能被接受。  

### Socket通信
套接字socket：网络中不同主机上的应用进程之间进行双向通信的端点。  
每台主机有一个唯一的主机地址标识（IP），同时主机内还有标识服务的序号id ，称作端口（port）。  
socket绑定了相应的IP和port，可以用（IP : port）的形式表示一个socket地址。  
一个连接由它两端的socket地址唯一确定：（ClientIP : ClientPort, ServerIP : ServerPort）   
当客户端发起一个连接请求时，客户端socket地址中的端口由系统自动分配，服务器端套接字地址中的端口通常是某个和服务相对应的知名端口。（例如Web服务器常使用端口80，电子邮件服务器使用端口25）  

信息：需要寄的快递  
IP：小区  
Port：门牌号,共有65536个端口  
Socket：快递地址（小区+门牌号）  
TCP，UCP等协议：快递公司  
利用socket发送消息：把快递（消息）放到门口（socket），由快递公司（TCP等协议）负责送到对应的地址（对方socket）  

Socket是传输层和应用层之间的软件抽象层，是一组接口。  
对于用户来说，socket把复杂的TCP/IP协议族隐藏在接口后，只需要遵循socket的规范，就能得到遵循TCP/UDP标准的程序。

创建套接字: socket.socket(family, type)  
参数说明：  
family：套接字家族，可以使AF_UNIX或者AF_INET，一般是AF_INET（IPv4）。  
type：套接字类型，根据是面向连接的还是非连接分为SOCK_STREAM（TCP）或SOCK_DGRAM（UDP），也就是TCP和UDP的区别，一般是SOCK_STREAM。  

连接流程：  
服务器端：初始化socket，与IP端口绑定，对IP端口进行监听，调用accept()阻塞，等待客户端连接。  
客户端：初始化socket，连接服务器。连接成功后客户端发送数据请求，服务器端接收并处理请求、回应数据，客户端读取数据。  
最后关闭连接，一次交互结束。  

```python
# 服务器端：
#   socket()->bind()->listen()->accept()->read()<=>write()->close()
# 客户端：                   |                  ||
#   socket()->       connect()->        write()<=>read()->close()

# 单线程服务端
# 弊端：顺序，一个客户端堵塞会影响其余客户端
import socket
import time

print("初始化服务器主机信息")
port = 5000       # 端口 0--1024为系统保留
host = '0.0.0.0'  # 监听所有可用网络接口
address = (host, port)
print("初始化服务器主机套接字对象socket()")
server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)  # 端口复用
print("绑定主机信息bind()")
server.bind(address)       # 绑定地址和端口
print("开始启动服务器listen()")
server.listen(5)           # 最大等待连接数为5
print("服务器启动，等待连接...")

while True:
    try:
        # 这里为了能停下来设置timeout
        server.settimeout(5.0)
        print("等待客户端连接")
        conn, addr = server.accept()      # 阻塞等待客户端连接
        print("客户端地址：", addr)
        conn.send("欢迎访问服务器".encode('utf-8'))
        time.sleep(1)                     # 模拟长时间处理（阻塞）
        print("处理完毕，关闭连接")
        conn.close()
    except socket.timeout:
        print("超时，关闭server")
        server.close()
        break
    except Exception as e:
        print(f"错误：{e}")
        break
```

```python
# 多线程服务端
import socket
import threading

def handle_client(conn, addr):
    print(f'新连接：{addr}')
    conn.send("欢迎访问多线程服务器".encode('utf-8'))
    conn.close()
    print(f'连接关闭：{addr}')

port = 5001
host = '0.0.0.0'
server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
server.bind((host, port))
server.listen(5)
print('服务器启动，等待连接...')

while True:
    try:
        # 这里为了能停下来设置timeout
        server.settimeout(5.0)
        print("等待客户端连接")
        conn, addr = server.accept()
        thread = threading.Thread(target=handle_client, args=(conn, addr))
        thread.start()
    except socket.timeout:
        print("超时，关闭server")
        server.close()
        break
    except Exception as e:
        print(f"错误：{e}")
        break

# 多进程服务端同理改成
# p = Process(target=handle_client,args=(conn,addr))
# p.start()
```

### The Pandas DataFrame Object
视角1：多个对齐的序列（series）的组合（record）  
视角2：支持多层索引的二维数据表  

可以输入给DataFrame构造器的数据：  
二维ndarray：数据矩阵，还可以传入行标和列标  
由数组、列表或元组组成的字典：每个序列会变成DataFrame的一列。所有序列的长度必须相同  
NumPy的结构化/记录数组：类似于“由数组组成的字典”  
由Series组成的字典：每个Series会成为一列。如果没有显式指定索引，则各Series的索引会被合并成结果的行索引

字典或Series的列表：各项将会成为DataFrame的一行。字典键或Series索引的并集将会成为DataFrame的列标  
由列表或元组组成的列表：类似于“二维ndarray”  
另一个DataFrame：该DataFrame的索引将会被沿用，除非显式指定了其他索引  
NumPy的MaskedArray：类似于“二维ndarray”的情况，只是掩码值在结果DataFrame会变成NA/缺失值

```python
# One-to-one joins：
df1 = pd.DataFrame({
    'employee': ['Bob', 'Jake', 'Lisa', 'Sue'],
    'group': ['Accounting', 'Engineering', 'Engineering', 'HR']
})
df2 = pd.DataFrame({
    'employee': ['Lisa', 'Bob', 'Jake', 'Sue'],
    'hire_date': [2004, 2008, 2012, 2014]
})

print("df1 (员工-部门):")
display(df1)
print("df2 (员工-入职日期):")
display(df2)

# 两个表有一个同名字段，自动按公共字段为轴进行合并
df3 = pd.merge(df1, df2)
print("merge(df1, df2) 自动合并结果:")
display(df3)

# 指定特定字段进行关联 merge-on
df3_explicit = pd.merge(df1, df2, on='employee')
print("pd.merge(df1, df2, on='employee'):")
display(df3_explicit)
```

```python
# left_on,right_on 可以指定不同名字段关联
df3_salary = pd.DataFrame({
    'name': ['Bob', 'Jake', 'Lisa', 'Sue'],
    'salary': [70000, 80000, 120000, 90000]
})

print("df1 (员工-部门):")
display(df1)
print("df3_salary (姓名-薪水):")
display(df3_salary)

# 左右表字段名不同时指定
merge_diff = pd.merge(df1, df3_salary, 
                      left_on='employee', 
                      right_on='name')
# 删除冗余列
merge_clean = merge_diff.drop('name', axis=1)
print("merge(df1, df3_salary, left_on='employee', right_on='name'):")
display(merge_clean)
```

```python
# Many-to-one joins：有重键值与对应的唯一键值进行join，单值对应的记录展开为多个
df4 = pd.DataFrame({
    'group': ['Accounting', 'Engineering', 'HR'],
    'supervisor': ['Carly', 'Guido', 'Steve']
})

print("df3 (员工-部门-入职日期):")
display(df3)
print("df4 (部门-主管):")
display(df4)

# 多对一合并（部门列有重复值）
many_to_one = pd.merge(df3, df4)
print("pd.merge(df3, df4) 多对一合并结果:")
display(many_to_one)
```

```python
# Many-to-many joins：多对多关联操作可以认为是一对多操作的复用
df5 = pd.DataFrame({
    'group': ['Accounting', 'Accounting',
              'Engineering', 'Engineering', 'HR', 'HR'],
    'skills': ['math', 'spreadsheets', 'coding', 'linux',
               'spreadsheets', 'organization']
})

print("df1 (员工-部门):")
display(df1)
print("df5 (部门-技能):")
display(df5)

# 多对多合并（部门列在两边都有重复）
many_to_many = pd.merge(df1, df5)
print("pd.merge(df1, df5) 多对多合并结果:")
display(many_to_many)
```

```python
# Inner-join 求索引交集
# outer-join是索引的并集（left join，right join…）
df6 = pd.DataFrame({
    'name': ['Peter', 'Paul', 'Mary'],
    'food': ['fish', 'beans', 'bread']
}, columns=['name', 'food'])

df7 = pd.DataFrame({
    'name': ['Mary', 'Joseph'],
    'drink': ['wine', 'beer']
}, columns=['name', 'drink'])

print("df6 (姓名-食物):")
display(df6)
print("df7 (姓名-饮料):")
display(df7)

# 内连接（默认how='inner'）
inner_join = pd.merge(df6, df7, how='inner')
print("pd.merge(df6, df7, how='inner') 内连接结果:")
display(inner_join)

# 外连接（并集）
outer_join = pd.merge(df6, df7, how='outer')
print("pd.merge(df6, df7, how='outer') 外连接结果:")
display(outer_join)

# 左连接
left_join = pd.merge(df6, df7, how='left')
print("左连接 (how='left'):")
display(left_join)

# 右连接
right_join = pd.merge(df6, df7, how='right')
print("右连接 (how='right'):")
display(right_join)
```

### Data Aggregations on Multi-Indices多索引数据聚合
Pandas内置了数据聚合方法，如mean()、sum()和max()。  
对于层次索引数据，这些方法可传递level参数，以控制聚合计算的数据子集。  
也可以按特定键分组 Group by certain Key。

```python
df = pd.DataFrame({
    'A': ['bob', 'john', 'bob', 'jeff', 'bob', 'jeff', 'bob', 'john'],
    'B': ['one', 'one', 'two', 'three', 'two', 'two', 'one', 'three'],
    'C': [3, 1, 4, 1, 5, 9, 2, 6],
    'D': [1, 2, 3, 4, 5, 6, 7, 8]
})
print("原始DataFrame:")
display(df)

# 基础分组聚合操作
g_count = df.groupby('A').count()
print("df.groupby('A').count():")
print(g_count)

g_sum = df.groupby('A').sum()
print("df.groupby('A').sum():")
print(g_sum)
```

```python
# apply 方法
grouped = df.groupby('A')
print("遍历每个分组:")
for name, group in grouped:
    print(f"组名: {name}")
    print(f"该组数据 ({len(group)} 行):")
    print(group)

# 在每个分组中取前2行
d_head = grouped.apply(lambda x: x.head(2))
print("grouped.apply(lambda x: x.head(2)):")
display(d_head)

# 自定义函数应用
def group_info(x):
    return pd.Series({
        'count': len(x),
        'C_sum': x['C'].sum(),
        'D_mean': x['D'].mean(),
        'B_unique': x['B'].nunique()
    })

group_stats = grouped.apply(group_info)
print("自定义函数 group_info 应用结果:")
display(group_stats)
```

```python
# agg()聚合函数为每个组返回单个聚合值。通过分组系列，还可以传递函数的列表或字典来进行聚合，并生成DataFrame。
import pandas as pd
import numpy as np

ipl_data = {
    'Team': ['Riders', 'Riders', 'Devils', 'Devils', 'Kings',
             'Kings', 'Kings', 'Kings', 'Riders', 'Royals', 'Royals', 'Riders'],
    'Rank': [1, 2, 3, 3, 4, 1, 1, 2, 4, 1, 2, 2],
    'Year': [2014, 2015, 2014, 2015, 2014, 2015, 2016, 2017, 2016, 2014, 2015, 2017],
    'Points': [876, 789, 863, 673, 741, 812, 756, 788, 694, 701, 804, 690]
}
df = pd.DataFrame(ipl_data)
print("原始DataFrame:")
display(df.head())

# 按Team分组
grouped = df.groupby('Team')
# 对Points列应用多个聚合函数
agg_points = grouped['Points'].agg(['sum', 'mean', 'std', 'min', 'max'])
print("Points列的多种聚合结果:")
display(agg_points)

# 对不同列应用不同聚合函数
agg_multi = grouped.agg({
    'Points': ['sum', 'max', 'mean'],
    'Rank': ['min', 'max', 'mean'],
    'Year': ['min', 'max', 'count']
})
print("多列不同函数聚合结果:")
display(agg_multi)

# 自定义聚合函数
def range_func(x):
    return x.max() - x.min()

agg_custom = grouped['Points'].agg([
    ('总和', 'sum'),
    ('平均值', 'mean'),
    ('标准差', 'std'),
    ('极差', range_func),
])
print("使用自定义函数的聚合结果:")
display(agg_custom)

# 使用named aggregation
agg_named = grouped.agg(
    平均分=('Points', 'mean'),
    最高分=('Points', 'max'),
    最低分=('Points', 'min'),
    比赛场次=('Points', 'count'),
    平均排名=('Rank', 'mean')
)
print("命名聚合结果:")
display(agg_named)
```
